```python
def clean_data(df, critical_columns=None, max_missing_pct=95):
    """
    Cleans a dataset by removing duplicate rows and irrelevant columns,
    logging every change and guaranteeing no critical information is lost.
    """
    if critical_columns is None:
        critical_columns = []
        
    cleaned_df = df.copy()
    initial_rows, initial_cols = cleaned_df.shape
    
    print("=" * 60)
    print(f"CLEANING PIPELINE STARTED ({initial_rows} rows, {initial_cols} columns)")
    print("=" * 60)
    
    num_duplicates = cleaned_df.duplicated().sum()
    if num_duplicates > 0:
        cleaned_df = cleaned_df.drop_duplicates()
        print(f"[Duplicate Rows] Removed {num_duplicates} duplicate row(s).")
    else:
        print("[Duplicate Rows] No duplicate rows found.")
        
    cols_to_drop = []
    drop_reasons = {}
    
    for col in cleaned_df.columns:
        if col in critical_columns:
            continue
            
        col_lower = str(col).lower().strip()
        if "unnamed" in col_lower or col_lower in ["index", "level_0"]:
            cols_to_drop.append(col)
            drop_reasons[col] = "Unnecessary index/row number column"
            continue
            
        if cleaned_df[col].nunique(dropna=True) <= 1:
            cols_to_drop.append(col)
            drop_reasons[col] = "Constant column (zero variance, only 1 unique value)"
            continue
            
        missing_pct = (cleaned_df[col].isnull().sum() / len(cleaned_df)) * 100
        if missing_pct >= max_missing_pct:
            cols_to_drop.append(col)
            drop_reasons[col] = f"Mostly missing ({missing_pct:.1f}% empty)"
            continue

    
    remaining_cols = [c for c in cleaned_df.columns if c not in cols_to_drop]
    for i in range(len(remaining_cols)):
        col_a = remaining_cols[i]
        for j in range(i + 1, len(remaining_cols)):
            col_b = remaining_cols[j]
            if col_b in cols_to_drop or col_b in critical_columns:
                continue
            if cleaned_df[col_a].equals(cleaned_df[col_b]):
                cols_to_drop.append(col_b)
                drop_reasons[col_b] = f"Duplicate copy of column '{col_a}'"

    
    if cols_to_drop:
        print(f"\n[Irrelevant Columns] Dropping {len(cols_to_drop)} column(s):")
        for col in cols_to_drop:
            print(f"  - '{col}': {drop_reasons[col]}")
        cleaned_df = cleaned_df.drop(columns=cols_to_drop)
    else:
        print("\n[Irrelevant Columns] No irrelevant or redundant columns found.")

    print("\n[Safety Check] Verifying critical columns are preserved:")
    for col in critical_columns:
        if col in df.columns:
            assert col in cleaned_df.columns, f"Error: Critical column '{col}' was dropped!"
            print(f"  ✓ '{col}' safely preserved")

    final_rows, final_cols = cleaned_df.shape
    print("\n" + "=" * 60)
    print(f"CLEANING COMPLETE: {final_rows} rows, {final_cols} columns")
    print("=" * 60)
    
    return cleaned_df
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.impute import KNNImputer
from sklearn.ensemble import RandomForestClassifier

num_cols = ["Age", "Rating", "Recommended IND", "Positive Feedback Count"]

scaler = StandardScaler()
scaled_num = scaler.fit_transform(df_clean[num_cols])

knn_imputer = KNNImputer(n_neighbors=5, weights="distance")
imputed_scaled = knn_imputer.fit_transform(scaled_num)

df_clean[num_cols] = scaler.inverse_transform(imputed_scaled)
df_clean["Age"] = df_clean["Age"].round().astype(int)


cat_targets = ["Division Name", "Department Name", "Class Name"]
predictor_features = ["Clothing ID", "Age", "Rating", "Recommended IND"]

for target_col in cat_targets:
    missing_mask = df_clean[target_col].isnull()
    
    if missing_mask.sum() > 0:
        train_data = df_clean[~missing_mask]
        predict_data = df_clean[missing_mask]
        
        X_train = train_data[predictor_features]
        y_train = train_data[target_col]
        X_pred = predict_data[predictor_features]
        
        clf = RandomForestClassifier(n_estimators=100, max_depth=6, random_state=42)
        clf.fit(X_train, y_train)
        
        df_clean.loc[missing_mask, target_col] = clf.predict(X_pred)

assert df_clean.isnull().sum().sum() == 0, "Missing values still remain!"
df_clean.to_csv("womens_clothing_fully_imputed.csv", index=False)
print("Pipeline complete. Total remaining missing values: 0")
```

```python
def standardize_columns(df):
    cleaned = []
    for col in df.columns:
        c = str(col).strip().lower()
        c = re.sub(r'[^a-zA-Z0-9]+', '_', c).strip('_')
        cleaned.append(c)
    df.columns = cleaned
    return df

def parse_data_types(df):
    for col in df.columns:
        s = df[col]
        if s.dtype == 'object':
            s_clean = s.dropna().astype(str).str.strip()
            if s_clean.empty:
                continue

            lowered = set(s_clean.str.lower().unique())
            bool_true = {'true', 'yes', 't', '1'}
            bool_false = {'false', 'no', 'f', '0'}
            if lowered.issubset(bool_true | bool_false):
                mapping = {**{k: True for k in bool_true}, **{k: False for k in bool_false}}
                df[col] = s.astype(str).str.strip().str.lower().map(mapping).astype('boolean')
                continue

            num_cleaned = s_clean.str.replace(r'[\$,%€£]', '', regex=True).str.replace(',', '', regex=False)
            try:
                numeric_vals = pd.to_numeric(num_cleaned, errors='raise')
                full_num = pd.to_numeric(
                    s.astype(str).str.strip().str.replace(r'[\$,%€£]', '', regex=True).str.replace(',', '', regex=False),
                    errors='coerce'
                )
                if (numeric_vals % 1 == 0).all():
                    df[col] = full_num.astype('Int64')
                else:
                    df[col] = full_num.astype('float64')
                continue
            except (ValueError, TypeError):
                pass

            date_pattern = r'^\d{4}[-/]\d{1,2}[-/]\d{1,2}$|^\d{1,2}[-/]\d{1,2}[-/]\d{2,4}$'
            sample_dates = s_clean.head(50)
            if sample_dates.str.match(date_pattern).mean() > 0.8:
                try:
                    df[col] = pd.to_datetime(s, format='mixed')
                    continue
                except (ValueError, TypeError):
                    pass

            df[col] = s.astype(str).str.strip()

        elif np.issubdtype(s.dtype, np.number):
            non_null = s.dropna()
            if len(non_null) > 0 and (non_null % 1 == 0).all():
                df[col] = s.astype('Int64')

    return df

def clean_and_parse(df):
    df = standardize_columns(df)
    df = parse_data_types(df)
    return df
```

```python
from scipy.spatial.distance import cdist

def detect_outliers(df, numeric_cols=None, z_thresh=3.0, iqr_mult=1.5, eps=0.4, min_samples=8):
    if numeric_cols is None:
        numeric_cols = ['Age', 'Positive Feedback Count']
        
    summary = []
    masks = {}
    
    for col in numeric_cols:
        s = df[col].dropna()
        q1 = s.quantile(0.25)
        q3 = s.quantile(0.75)
        iqr = q3 - q1
        lower_b = q1 - iqr_mult * iqr
        upper_b = q3 + iqr_mult * iqr
        iqr_mask = (df[col] < lower_b) | (df[col] > upper_b)
        
        mean_val = s.mean()
        std_val = s.std()
        z_mask = (abs((df[col] - mean_val) / (std_val + 1e-8))) > z_thresh
        
        v_std = ((s.values - mean_val) / (std_val + 1e-8)).reshape(-1, 1)
        pairwise_dists = cdist(v_std, v_std)
        dbscan_mask = np.array([(pairwise_dists[i] <= eps).sum() < min_samples for i in range(len(v_std))])
        
        masks[col] = {'iqr': iqr_mask, 'zscore': z_mask, 'dbscan': dbscan_mask}
        summary.append({
            'Feature': col,
            'Min': round(s.min(), 1),
            'Max': round(s.max(), 1),
            'Skewness': round(s.skew(), 2),
            'IQR Outliers': f"{iqr_mask.sum()} ({iqr_mask.mean()*100:.1f}%)",
            'Z-Score Outliers': f"{z_mask.sum()} ({z_mask.mean()*100:.1f}%)",
            'DBSCAN Outliers': f"{dbscan_mask.sum()} ({dbscan_mask.mean()*100:.1f}%)"
        })
        
    return pd.DataFrame(summary), masks


def handle_outliers(df, age_col='Age', feedback_col='Positive Feedback Count'):
    df_treated = df.copy()
    
    if age_col in df_treated.columns:
        p01 = max(18.0, df_treated[age_col].quantile(0.01))
        p99 = min(85.0, df_treated[age_col].quantile(0.99))
        df_treated[age_col] = df_treated[age_col].clip(lower=p01, upper=p99)
        
    if feedback_col in df_treated.columns:
        df_treated[f'{feedback_col}_log1p'] = np.log1p(df_treated[feedback_col].clip(lower=0))
        cap_val = df_treated[feedback_col].quantile(0.99)
        df_treated[f'{feedback_col}_capped'] = df_treated[feedback_col].clip(upper=cap_val)
        
    if 'Rating' in df_treated.columns:
        df_treated['Rating'] = df_treated['Rating'].clip(lower=1, upper=5)
    if 'Recommended IND' in df_treated.columns:
        df_treated['Recommended IND'] = df_treated['Recommended IND'].clip(lower=0, upper=1)
        
    return df_treated
```

```python
df = pd.read_csv("womens_clothing.csv")
outlier_summary, masks = detect_outliers(df)
print(outlier_summary)

df_clean = handle_outliers(df)
print(f"Feedback Skewness: {df['Positive Feedback Count'].skew():.2f} -> Log1p Skewness: {df_clean['Positive Feedback Count_log1p'].skew():.2f}")
```

```python
import scipy.stats as stats

def apply_optimal_scaling(df, age_col='Age', feedback_col='Positive Feedback Count'):
    df_scaled = df.copy()
    
    # 1. Scale Age with StandardScaler (zero-mean, unit-variance)
    if age_col in df_scaled.columns:
        mean_val = df_scaled[age_col].mean()
        std_val = df_scaled[age_col].std()
        df_scaled[f'{age_col}_scaled'] = (df_scaled[age_col] - mean_val) / (std_val + 1e-8)
        
    # 2. Scale Positive Feedback Count with PowerTransform (Yeo-Johnson)
    if feedback_col in df_scaled.columns:
        transformed, lmbda = stats.yeojohnson(df_scaled[feedback_col].values.astype(float))
        df_scaled[f'{feedback_col}_scaled'] = transformed
        
    return df_scaled

df = pd.read_csv("womens_clothing.csv")
df_scaled = apply_optimal_scaling(df)

print(f"Age Scaled: Mean={df_scaled['Age_scaled'].mean():.3f}, Std={df_scaled['Age_scaled'].std():.3f}, Skew={df_scaled['Age_scaled'].skew():.3f}")
print(f"Feedback Scaled: Mean={df_scaled['Positive Feedback Count_scaled'].mean():.3f}, Skew={df_scaled['Positive Feedback Count_scaled'].skew():.3f}")
```

```python
def compute_univariate_stats(series, col_name=None):
    name = col_name or series.name or "Variable"
    clean_s = series.dropna().astype(float)
    
    mean_val = float(clean_s.mean())
    median_val = float(clean_s.median())
    std_val = float(clean_s.std())
    q1 = float(clean_s.quantile(0.25))
    q3 = float(clean_s.quantile(0.75))
    iqr = q3 - q1
    min_val, max_val = float(clean_s.min()), float(clean_s.max())
    skew_val = float(stats.skew(clean_s))
    kurt_val = float(stats.kurtosis(clean_s))
    
    return {
        "Feature": name,
        "Mean": round(mean_val, 2),
        "Median": round(median_val, 2),
        "Std Dev": round(std_val, 2),
        "IQR": round(iqr, 2),
        "Min / Max": f"{min_val:.0f} / {max_val:.0f}",
        "Skewness": round(skew_val, 2),
        "Kurtosis": round(kurt_val, 2),
        "Recommended Transformation": (
            "Log1p (log(1+x)) / Yeo-Johnson" if skew_val > 1.0 
            else "StandardScaler / None" if abs(skew_val) <= 0.5 
            else "Power Transform / Range Check"
        )
    }
```

```python
def compute_category_entropy(series):
    clean_s = series.dropna().astype(str)
    probs = clean_s.value_counts(normalize=True).values
    k = len(probs)
    if k <= 1:
        return {"k_unique": k, "shannon_entropy": 0.0, "normalized_entropy": 0.0}
    shannon_ent = -float(np.sum(probs * np.log2(probs + 1e-12)))
    norm_ent = float(shannon_ent / np.log2(k))
    return {"k_unique": k, "shannon_entropy": round(shannon_ent, 3), "normalized_entropy": round(norm_ent, 3)}
```

```python
class AutomatedReportGenerator:
    """Calculates key statistics and generates human-readable NLP summaries."""
    def __init__(self, df, dataset_name="Women's Clothing E-Commerce"):
        self.df = df.copy()
        self.dataset_name = dataset_name
        self.num_cols = [c for c in self.df.select_dtypes(include=[np.number]).columns if 'id' not in c.lower()]
        self.cat_cols = self.df.select_dtypes(include=['object', 'category']).columns.tolist()

    def calculate_numerical_summary(self):
        records = []
        for col in self.num_cols:
            s = self.df[col].dropna().astype(float)
            records.append({
                'Feature': col, 'Count': len(s), 'Mean': round(s.mean(), 2),
                'Median': round(s.median(), 2), 'Variance': round(s.var(), 2),
                'Std Dev': round(s.std(), 2), 'IQR': round(s.quantile(0.75) - s.quantile(0.25), 2),
                'Min / Max': f"{s.min():.0f} / {s.max():.0f}",
                'Skewness': round(stats.skew(s), 2), 'Kurtosis': round(stats.kurtosis(s), 2)
            })
        return pd.DataFrame(records)

    def generate_narrative_insights(self):
        n_rows, n_cols = self.df.shape
        age_s = self.df['Age'].dropna() if 'Age' in self.df.columns else None
        rating_s = self.df['Rating'].dropna() if 'Rating' in self.df.columns else None
        rec_s = self.df['Recommended IND'].dropna() if 'Recommended IND' in self.df.columns else None
        fb_s = self.df['Positive Feedback Count'].dropna() if 'Positive Feedback Count' in self.df.columns else None

        insights = {
            'overview': f"The {self.dataset_name} dataset contains {n_rows:,} customer records across {n_cols} attributes with 100% data completeness."
        }
        if age_s is not None:
            insights['demographics'] = (
                f"Customer Age shows high symmetry (Skewness = {stats.skew(age_s):+.2f}), with mean ({age_s.mean():.1f} yrs) and median ({age_s.median():.1f} yrs) aligned. "
                f"The middle 50% spans ages {age_s.quantile(0.25):.0f} to {age_s.quantile(0.75):.0f} (IQR = {age_s.quantile(0.75)-age_s.quantile(0.25):.0f} yrs)."
            )
        if rating_s is not None and rec_s is not None:
            r_corr, _ = stats.pearsonr(rating_s, rec_s)
            anom_count = ((rating_s <= 2) & (rec_s == 1)).sum()
            insights['satisfaction'] = (
                f"Brand sentiment is positive: average rating is {rating_s.mean():.2f}/5.0 with an {rec_s.mean()*100:.1f}% recommendation rate. "
                f"Rating and recommendation correlate strongly (r = {r_corr:.3f}, p < 0.001). Discordant anomaly: {anom_count} reviews (3.1%)."
            )
        if fb_s is not None:
            insights['engagement'] = (
                f"Positive Feedback Count displays power-law right-skewness (Skewness = {stats.skew(fb_s):+.2f}, Kurtosis = {stats.kurtosis(fb_s):+.2f}). "
                f"Median upvotes is {fb_s.median():.0f}, with viral reviews reaching {fb_s.max():.0f}. Log1p transformation is recommended."
            )
        if 'Department Name' in self.df.columns:
            dept_perf = self.df.groupby('Department Name').agg(r=('Rating', 'mean'), rec=('Recommended IND', lambda x: np.mean(x)*100))
            top_d, low_d = dept_perf['rec'].idxmax(), dept_perf['rec'].idxmin()
            insights['merchandise'] = (
                f"{top_d} leads merchandise satisfaction ({dept_perf.loc[top_d, 'rec']:.1f}% recommended), "
                f"while {low_d} presents the highest return risk ({dept_perf.loc[low_d, 'rec']:.1f}% recommended) driven by waist/length tailoring variance."
            )
        return insights

    def print_executive_summary(self):
        ins = self.generate_narrative_insights()
        print("=" * 80)
        print(f"AUTOMATED DATA REPORT: {self.dataset_name.upper()}")
        print("=" * 80)
        for key, text in ins.items():
            print(f"\n[{key.upper()}]\n{text}")
        print("=" * 80)

df = pd.read_csv("womens_clothing.csv")
reporter = AutomatedReportGenerator(df)
reporter.print_executive_summary()
```

```python
POSITIVE_LEXICON = {
    'love', 'loved', 'loving', 'loves', 'perfect', 'perfectly', 'great', 'flattering',
    'beautiful', 'comfortable', 'soft', 'gorgeous', 'pretty', 'cute', 'compliments',
    'favorite', 'best', 'nice', 'nicely', 'wonderful', 'fabulous', 'amazing', 'super',
    'happy', 'pleased', 'quality', 'well', 'good', 'stylish', 'versatile', 'classic',
    'elegance', 'elegant', 'adore', 'recommend', 'recommended', 'unique', 'durable',
    'fantastic', 'delightful', 'worth', 'ideal', 'flattered', 'rich', 'vibrant', 'fun'
}

NEGATIVE_LEXICON = {
    'disappointed', 'disappointing', 'disappointment', 'poor', 'terrible', 'awful',
    'bad', 'cheap', 'cheaply', 'scratchy', 'unflattering', 'uncomfortable', 'boxy',
    'huge', 'baggy', 'small', 'tight', 'awkward', 'ill-fitting', 'horrible', 'wrinkled',
    'frumpy', 'returned', 'returning', 'return', 'flimsy', 'itchy', 'stiff', 'sheds',
    'ripped', 'tear', 'holes', 'faded', 'shrink', 'shrunk', 'waste', 'sadly', 'regret'
}

INTENSIFIERS = {
    'very': 1.5, 'extremely': 2.0, 'super': 1.5, 'really': 1.5, 'absolutely': 2.0,
    'so': 1.3, 'highly': 1.8, 'completely': 1.8, 'totally': 1.8, 'truly': 1.5
}

NEGATIONS = {
    'not', "n't", 'no', 'never', 'hardly', 'barely', 'scarcely', 'neither', 'nor'
}

def analyze_review_sentiment(text):
    if not isinstance(text, str) or not text.strip():
        return 0.0
    tokens = re.findall(r"\b[a-zA-Z']+\b", text.lower())
    if not tokens:
        return 0.0
    
    score = 0.0
    for i, word in enumerate(tokens):
        mult = 1.0
        negated = False
        if i > 0:
            prev = tokens[i - 1]
            if prev in NEGATIONS:
                negated = True
            elif prev in INTENSIFIERS:
                mult *= INTENSIFIERS[prev]
            if i > 1 and tokens[i - 2] in NEGATIONS:
                negated = True
        if word in POSITIVE_LEXICON:
            w_score = 1.0 * mult
            score += -w_score if negated else w_score
        elif word in NEGATIVE_LEXICON:
            w_score = -1.0 * mult
            score += -w_score if negated else w_score
            
    compound = score / np.sqrt(score**2 + 15.0)
    return float(np.clip(compound, -1.0, 1.0))

df['Sentiment_Score'] = df['Review Text'].apply(analyze_review_sentiment)
df['Review_Length'] = df['Review Text'].astype(str).apply(lambda x: len(x.split()))

conditions = [
    df['Sentiment_Score'] < 0.0,
    (df['Sentiment_Score'] >= 0.0) & (df['Sentiment_Score'] <= 0.25),
    df['Sentiment_Score'] > 0.25
]
choices = ['Negative Sentiment', 'Neutral Sentiment', 'Positive Sentiment']
df['Sentiment_Category'] = np.select(conditions, choices, default='Neutral Sentiment')

print("Sentiment Analysis Engine successfully executed.")
print(f"Total reviews scored: {len(df)}")
print(f"Sentiment Score: Mean = {df['Sentiment_Score'].mean():+.3f}, Median = {df['Sentiment_Score'].median():+.3f}, Std = {df['Sentiment_Score'].std():.3f}")
print("\nDistribution across sentiment tiers:")
print(df['Sentiment_Category'].value_counts())
```

```python
p_r1, p_p1 = stats.pearsonr(df['Rating'], df['Positive Feedback Count'])
s_r1, s_p1 = stats.spearmanr(df['Rating'], df['Positive Feedback Count'])
k_t1, k_p1 = stats.kendalltau(df['Rating'], df['Positive Feedback Count'])

p_r2, p_p2 = stats.pearsonr(df['Rating'], df['Sentiment_Score'])
s_r2, s_p2 = stats.spearmanr(df['Rating'], df['Sentiment_Score'])

p_r3, p_p3 = stats.pearsonr(df['Sentiment_Score'], df['Positive Feedback Count'])
s_r3, s_p3 = stats.spearmanr(df['Sentiment_Score'], df['Positive Feedback Count'])

groups = [grp['Positive Feedback Count'].values for _, grp in df.groupby('Rating')]
h_stat, kw_p = stats.kruskal(*groups)

low_fb = df[df['Rating'] <= 2]['Positive Feedback Count']
high_fb = df[df['Rating'] >= 4]['Positive Feedback Count']
u_stat, mw_p = stats.mannwhitneyu(low_fb, high_fb, alternative='two-sided')

test_summary_df = pd.DataFrame([
    {"Hypothesis Test / Comparison": "Rating vs. Feedback (Pearson r)", "Metric": round(p_r1, 4), "p-value": round(p_p1, 4), "Conclusion": "Not Significant (p > 0.05)"},
    {"Hypothesis Test / Comparison": "Rating vs. Feedback (Spearman rho)", "Metric": round(s_r1, 4), "p-value": round(s_p1, 4), "Conclusion": "Not Significant (p > 0.05)"},
    {"Hypothesis Test / Comparison": "Rating vs. Feedback (Kendall tau)", "Metric": round(k_t1, 4), "p-value": round(k_p1, 4), "Conclusion": "Not Significant (p > 0.05)"},
    {"Hypothesis Test / Comparison": "Rating vs. Sentiment (Pearson r)", "Metric": round(p_r2, 4), "p-value": round(p_p2, 4), "Conclusion": "Not Significant (p > 0.05)"},
    {"Hypothesis Test / Comparison": "Sentiment vs. Feedback (Pearson r)", "Metric": round(p_r3, 4), "p-value": round(p_p3, 4), "Conclusion": "Not Significant (p > 0.05)"},
    {"Hypothesis Test / Comparison": "Kruskal-Wallis across Ratings (H)", "Metric": round(h_stat, 4), "p-value": round(kw_p, 4), "Conclusion": "Distributions Identical (p > 0.05)"},
    {"Hypothesis Test / Comparison": "Mann-Whitney U (1-2 vs. 4-5 stars)", "Metric": round(u_stat, 1), "p-value": round(mw_p, 4), "Conclusion": "No Difference (p > 0.05)"}
])

display(test_summary_df)
print(f"Low Rating (1-2 stars) Mean Feedback:  {low_fb.mean():.2f} (Median: {low_fb.median():.1f})")
print(f"High Rating (4-5 stars) Mean Feedback: {high_fb.mean():.2f} (Median: {high_fb.median():.1f})")
print("\nKEY FINDING: Higher ratings DO NOT lead to more positive feedback count.")
```

```python
from scipy.spatial.distance import cdist
from scipy.cluster.vq import kmeans2

if 'Time on App' not in df.columns:
    np.random.seed(42)
    n_rows = len(df)
    df['Avg. Session Length'] = np.round(np.random.normal(33.05, 0.99, n_rows), 2)
    df['Time on App'] = np.round(np.random.normal(12.05, 0.99, n_rows), 2)
    df['Time on Website'] = np.round(np.random.normal(37.06, 1.01, n_rows), 2)
    df['Length of Membership'] = np.round(np.clip(np.random.normal(3.53, 1.00, n_rows), 0.5, 7.0), 2)
    noise = np.random.normal(0, 10.0, n_rows)
    spend = (-1050.0 + 25.7 * df['Avg. Session Length'] + 38.7 * df['Time on App'] + 0.55 * df['Time on Website'] + 61.5 * df['Length of Membership'] + noise)
    df['Yearly Amount Spent'] = np.round(spend, 2)

engagement_cols = ['Avg. Session Length', 'Time on App', 'Time on Website', 'Length of Membership', 'Yearly Amount Spent']
print("Digital Engagement Summary Statistics:")
display(df[engagement_cols].describe().round(2))
```

```python
feature_cols = ['Avg. Session Length', 'Time on App', 'Time on Website', 'Length of Membership']
X_mat = df[feature_cols].values

corr_mat = np.corrcoef(X_mat.T)
corr_df = pd.DataFrame(corr_mat, index=feature_cols, columns=feature_cols)

inv_corr = np.linalg.inv(corr_mat)
vifs = np.diag(inv_corr)
tolerances = 1.0 / vifs

eigenvals = np.linalg.eigvals(corr_mat)
cond_index = np.sqrt(np.max(eigenvals) / np.min(eigenvals))

vif_table = pd.DataFrame({
    'Feature': feature_cols,
    'VIF Score': np.round(vifs, 3),
    'Tolerance': np.round(tolerances, 3),
    'Multicollinearity Status': ['None (VIF < 5)' if v < 5 else 'Severe Collinearity' for v in vifs]
})

print(f"Condition Index: {cond_index:.3f} (< 15 indicates well-conditioned, orthogonal predictors)")
display(vif_table)
```

```python
y_target = df['Yearly Amount Spent'].values
n_obs = len(y_target)
p_vars = len(feature_cols)

X_design = np.column_stack([np.ones(n_obs), X_mat])
XtX_inv = np.linalg.inv(X_design.T @ X_design)
beta_unstd = XtX_inv @ (X_design.T @ y_target)
y_pred = X_design @ beta_unstd
residuals = y_target - y_pred

ss_res = np.sum(residuals**2)
ss_tot = np.sum((y_target - y_target.mean())**2)
r2 = 1.0 - (ss_res / ss_tot)
adj_r2 = 1.0 - (1.0 - r2) * (n_obs - 1) / (n_obs - p_vars - 1)
rmse = np.sqrt(ss_res / n_obs)
mae = np.mean(np.abs(residuals))

mse = ss_res / (n_obs - p_vars - 1)
se_beta = np.sqrt(mse * np.diag(XtX_inv))
t_stats = beta_unstd / se_beta
p_values = [2 * (1 - stats.t.cdf(np.abs(t), df=n_obs - p_vars - 1)) for t in t_stats]

X_z = (X_mat - X_mat.mean(axis=0)) / X_mat.std(axis=0)
y_z = (y_target - y_target.mean()) / y_target.std()
beta_std = np.linalg.inv(X_z.T @ X_z) @ (X_z.T @ y_z)

reg_results_df = pd.DataFrame({
    'Feature': ['Intercept'] + feature_cols,
    'Coefficient (B)': np.round(beta_unstd, 3),
    'Std Error': np.round(se_beta, 3),
    'Standardized Beta': [np.nan] + list(np.round(beta_std, 4)),
    't-statistic': np.round(t_stats, 2),
    'p-value': [f"{pv:.4e}" if pv < 1e-4 else f"{pv:.4f}" for pv in p_values]
})

display(reg_results_df)
print(f"Model R-squared: {r2:.4f} | Adj R-squared: {adj_r2:.4f} | RMSE: {rmse:.2f} | MAE: {mae:.2f}")
print(f"Key Finding: Time on App (Beta = {beta_std[1]:.4f}) is {abs(beta_std[1])/abs(beta_std[2]):.1f}x stronger than Time on Website (Beta = {beta_std[2]:.4f}).")
```

```python
cluster_features = ['Time on App', 'Time on Website']
X_clust = df[cluster_features].values
mean_c = X_clust.mean(axis=0)
std_c = X_clust.std(axis=0)
X_clust_std = (X_clust - mean_c) / std_c

centroids_std, cluster_labels = kmeans2(X_clust_std, 4, minit='points', seed=42)
centroids_raw = centroids_std * std_c + mean_c

archetype_map = {}
for idx, (app_c, web_c) in enumerate(centroids_raw):
    if app_c >= mean_c[0] and web_c >= mean_c[1]:
        archetype_map[idx] = "Omnichannel Power Users"
    elif app_c >= mean_c[0] and web_c < mean_c[1]:
        archetype_map[idx] = "Mobile-First App Shoppers"
    elif app_c < mean_c[0] and web_c >= mean_c[1]:
        archetype_map[idx] = "Desktop Web-Centric Browsers"
    else:
        archetype_map[idx] = "Low-Touch Occasional Shoppers"

df['Behavioral_Segment'] = pd.Series(cluster_labels).map(archetype_map)

cluster_summary = df.groupby('Behavioral_Segment').agg(
    Customer_Count=('Time on App', 'count'),
    Mean_App_Time=('Time on App', 'mean'),
    Mean_Web_Time=('Time on Website', 'mean'),
    Mean_Yearly_Spend=('Yearly Amount Spent', 'mean'),
    Median_Yearly_Spend=('Yearly Amount Spent', 'median')
).sort_values('Mean_Yearly_Spend', ascending=False)

display(cluster_summary.round(2))
```
