```python
def clean_data(df, critical_columns=None, max_missing_pct=95):
    """
    Cleans a dataset by removing duplicate rows and irrelevant columns,
    logging every change and guaranteeing no critical information is lost.
    """
    if critical_columns is None:
        critical_columns = []
        
    cleaned_df = df.copy()
    initial_rows, initial_cols = cleaned_df.shape
    
    print("=" * 60)
    print(f"CLEANING PIPELINE STARTED ({initial_rows} rows, {initial_cols} columns)")
    print("=" * 60)
    
    num_duplicates = cleaned_df.duplicated().sum()
    if num_duplicates > 0:
        cleaned_df = cleaned_df.drop_duplicates()
        print(f"[Duplicate Rows] Removed {num_duplicates} duplicate row(s).")
    else:
        print("[Duplicate Rows] No duplicate rows found.")
        
    cols_to_drop = []
    drop_reasons = {}
    
    for col in cleaned_df.columns:
        if col in critical_columns:
            continue
            
        col_lower = str(col).lower().strip()
        if "unnamed" in col_lower or col_lower in ["index", "level_0"]:
            cols_to_drop.append(col)
            drop_reasons[col] = "Unnecessary index/row number column"
            continue
            
        if cleaned_df[col].nunique(dropna=True) <= 1:
            cols_to_drop.append(col)
            drop_reasons[col] = "Constant column (zero variance, only 1 unique value)"
            continue
            
        missing_pct = (cleaned_df[col].isnull().sum() / len(cleaned_df)) * 100
        if missing_pct >= max_missing_pct:
            cols_to_drop.append(col)
            drop_reasons[col] = f"Mostly missing ({missing_pct:.1f}% empty)"
            continue

    
    remaining_cols = [c for c in cleaned_df.columns if c not in cols_to_drop]
    for i in range(len(remaining_cols)):
        col_a = remaining_cols[i]
        for j in range(i + 1, len(remaining_cols)):
            col_b = remaining_cols[j]
            if col_b in cols_to_drop or col_b in critical_columns:
                continue
            if cleaned_df[col_a].equals(cleaned_df[col_b]):
                cols_to_drop.append(col_b)
                drop_reasons[col_b] = f"Duplicate copy of column '{col_a}'"

    
    if cols_to_drop:
        print(f"\n[Irrelevant Columns] Dropping {len(cols_to_drop)} column(s):")
        for col in cols_to_drop:
            print(f"  - '{col}': {drop_reasons[col]}")
        cleaned_df = cleaned_df.drop(columns=cols_to_drop)
    else:
        print("\n[Irrelevant Columns] No irrelevant or redundant columns found.")

    print("\n[Safety Check] Verifying critical columns are preserved:")
    for col in critical_columns:
        if col in df.columns:
            assert col in cleaned_df.columns, f"Error: Critical column '{col}' was dropped!"
            print(f"  ✓ '{col}' safely preserved")

    final_rows, final_cols = cleaned_df.shape
    print("\n" + "=" * 60)
    print(f"CLEANING COMPLETE: {final_rows} rows, {final_cols} columns")
    print("=" * 60)
    
    return cleaned_df
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.impute import KNNImputer
from sklearn.ensemble import RandomForestClassifier

num_cols = ["Age", "Rating", "Recommended IND", "Positive Feedback Count"]

scaler = StandardScaler()
scaled_num = scaler.fit_transform(df_clean[num_cols])

knn_imputer = KNNImputer(n_neighbors=5, weights="distance")
imputed_scaled = knn_imputer.fit_transform(scaled_num)

df_clean[num_cols] = scaler.inverse_transform(imputed_scaled)
df_clean["Age"] = df_clean["Age"].round().astype(int)


cat_targets = ["Division Name", "Department Name", "Class Name"]
predictor_features = ["Clothing ID", "Age", "Rating", "Recommended IND"]

for target_col in cat_targets:
    missing_mask = df_clean[target_col].isnull()
    
    if missing_mask.sum() > 0:
        train_data = df_clean[~missing_mask]
        predict_data = df_clean[missing_mask]
        
        X_train = train_data[predictor_features]
        y_train = train_data[target_col]
        X_pred = predict_data[predictor_features]
        
        clf = RandomForestClassifier(n_estimators=100, max_depth=6, random_state=42)
        clf.fit(X_train, y_train)
        
        df_clean.loc[missing_mask, target_col] = clf.predict(X_pred)

assert df_clean.isnull().sum().sum() == 0, "Missing values still remain!"
df_clean.to_csv("womens_clothing_fully_imputed.csv", index=False)
print("Pipeline complete. Total remaining missing values: 0")
```

```python
def standardize_columns(df):
    cleaned = []
    for col in df.columns:
        c = str(col).strip().lower()
        c = re.sub(r'[^a-zA-Z0-9]+', '_', c).strip('_')
        cleaned.append(c)
    df.columns = cleaned
    return df

def parse_data_types(df):
    for col in df.columns:
        s = df[col]
        if s.dtype == 'object':
            s_clean = s.dropna().astype(str).str.strip()
            if s_clean.empty:
                continue

            lowered = set(s_clean.str.lower().unique())
            bool_true = {'true', 'yes', 't', '1'}
            bool_false = {'false', 'no', 'f', '0'}
            if lowered.issubset(bool_true | bool_false):
                mapping = {**{k: True for k in bool_true}, **{k: False for k in bool_false}}
                df[col] = s.astype(str).str.strip().str.lower().map(mapping).astype('boolean')
                continue

            num_cleaned = s_clean.str.replace(r'[\$,%€£]', '', regex=True).str.replace(',', '', regex=False)
            try:
                numeric_vals = pd.to_numeric(num_cleaned, errors='raise')
                full_num = pd.to_numeric(
                    s.astype(str).str.strip().str.replace(r'[\$,%€£]', '', regex=True).str.replace(',', '', regex=False),
                    errors='coerce'
                )
                if (numeric_vals % 1 == 0).all():
                    df[col] = full_num.astype('Int64')
                else:
                    df[col] = full_num.astype('float64')
                continue
            except (ValueError, TypeError):
                pass

            date_pattern = r'^\d{4}[-/]\d{1,2}[-/]\d{1,2}$|^\d{1,2}[-/]\d{1,2}[-/]\d{2,4}$'
            sample_dates = s_clean.head(50)
            if sample_dates.str.match(date_pattern).mean() > 0.8:
                try:
                    df[col] = pd.to_datetime(s, format='mixed')
                    continue
                except (ValueError, TypeError):
                    pass

            df[col] = s.astype(str).str.strip()

        elif np.issubdtype(s.dtype, np.number):
            non_null = s.dropna()
            if len(non_null) > 0 and (non_null % 1 == 0).all():
                df[col] = s.astype('Int64')

    return df

def clean_and_parse(df):
    df = standardize_columns(df)
    df = parse_data_types(df)
    return df
```

```python
from scipy.spatial.distance import cdist

def detect_outliers(df, numeric_cols=None, z_thresh=3.0, iqr_mult=1.5, eps=0.4, min_samples=8):
    if numeric_cols is None:
        numeric_cols = ['Age', 'Positive Feedback Count']
        
    summary = []
    masks = {}
    
    for col in numeric_cols:
        s = df[col].dropna()
        q1 = s.quantile(0.25)
        q3 = s.quantile(0.75)
        iqr = q3 - q1
        lower_b = q1 - iqr_mult * iqr
        upper_b = q3 + iqr_mult * iqr
        iqr_mask = (df[col] < lower_b) | (df[col] > upper_b)
        
        mean_val = s.mean()
        std_val = s.std()
        z_mask = (abs((df[col] - mean_val) / (std_val + 1e-8))) > z_thresh
        
        v_std = ((s.values - mean_val) / (std_val + 1e-8)).reshape(-1, 1)
        pairwise_dists = cdist(v_std, v_std)
        dbscan_mask = np.array([(pairwise_dists[i] <= eps).sum() < min_samples for i in range(len(v_std))])
        
        masks[col] = {'iqr': iqr_mask, 'zscore': z_mask, 'dbscan': dbscan_mask}
        summary.append({
            'Feature': col,
            'Min': round(s.min(), 1),
            'Max': round(s.max(), 1),
            'Skewness': round(s.skew(), 2),
            'IQR Outliers': f"{iqr_mask.sum()} ({iqr_mask.mean()*100:.1f}%)",
            'Z-Score Outliers': f"{z_mask.sum()} ({z_mask.mean()*100:.1f}%)",
            'DBSCAN Outliers': f"{dbscan_mask.sum()} ({dbscan_mask.mean()*100:.1f}%)"
        })
        
    return pd.DataFrame(summary), masks


def handle_outliers(df, age_col='Age', feedback_col='Positive Feedback Count'):
    df_treated = df.copy()
    
    if age_col in df_treated.columns:
        p01 = max(18.0, df_treated[age_col].quantile(0.01))
        p99 = min(85.0, df_treated[age_col].quantile(0.99))
        df_treated[age_col] = df_treated[age_col].clip(lower=p01, upper=p99)
        
    if feedback_col in df_treated.columns:
        df_treated[f'{feedback_col}_log1p'] = np.log1p(df_treated[feedback_col].clip(lower=0))
        cap_val = df_treated[feedback_col].quantile(0.99)
        df_treated[f'{feedback_col}_capped'] = df_treated[feedback_col].clip(upper=cap_val)
        
    if 'Rating' in df_treated.columns:
        df_treated['Rating'] = df_treated['Rating'].clip(lower=1, upper=5)
    if 'Recommended IND' in df_treated.columns:
        df_treated['Recommended IND'] = df_treated['Recommended IND'].clip(lower=0, upper=1)
        
    return df_treated
```

```python
df = pd.read_csv("womens_clothing.csv")
outlier_summary, masks = detect_outliers(df)
print(outlier_summary)

df_clean = handle_outliers(df)
print(f"Feedback Skewness: {df['Positive Feedback Count'].skew():.2f} -> Log1p Skewness: {df_clean['Positive Feedback Count_log1p'].skew():.2f}")
```

```python
import scipy.stats as stats

def apply_optimal_scaling(df, age_col='Age', feedback_col='Positive Feedback Count'):
    df_scaled = df.copy()
    
    # 1. Scale Age with StandardScaler (zero-mean, unit-variance)
    if age_col in df_scaled.columns:
        mean_val = df_scaled[age_col].mean()
        std_val = df_scaled[age_col].std()
        df_scaled[f'{age_col}_scaled'] = (df_scaled[age_col] - mean_val) / (std_val + 1e-8)
        
    # 2. Scale Positive Feedback Count with PowerTransform (Yeo-Johnson)
    if feedback_col in df_scaled.columns:
        transformed, lmbda = stats.yeojohnson(df_scaled[feedback_col].values.astype(float))
        df_scaled[f'{feedback_col}_scaled'] = transformed
        
    return df_scaled

# Execution
df = pd.read_csv("womens_clothing.csv")
df_scaled = apply_optimal_scaling(df)

print(f"Age Scaled: Mean={df_scaled['Age_scaled'].mean():.3f}, Std={df_scaled['Age_scaled'].std():.3f}, Skew={df_scaled['Age_scaled'].skew():.3f}")
print(f"Feedback Scaled: Mean={df_scaled['Positive Feedback Count_scaled'].mean():.3f}, Skew={df_scaled['Positive Feedback Count_scaled'].skew():.3f}")
```

```python
import matplotlib.pyplot as plt


def compute_univariate_stats(series, col_name=None):
    name = col_name or series.name or "Variable"
    clean_s = series.dropna().astype(float)
    
    mean_val = float(clean_s.mean())
    median_val = float(clean_s.median())
    std_val = float(clean_s.std())
    q1 = float(clean_s.quantile(0.25))
    q3 = float(clean_s.quantile(0.75))
    iqr = q3 - q1
    min_val, max_val = float(clean_s.min()), float(clean_s.max())
    skew_val = float(stats.skew(clean_s))
    kurt_val = float(stats.kurtosis(clean_s))
    
    return {
        "Feature": name,
        "Mean": round(mean_val, 2),
        "Median": round(median_val, 2),
        "Std Dev": round(std_val, 2),
        "IQR": round(iqr, 2),
        "Min / Max": f"{min_val:.0f} / {max_val:.0f}",
        "Skewness": round(skew_val, 2),
        "Kurtosis": round(kurt_val, 2),
        "Recommended Transformation": (
            "Log1p (log(1+x)) / Yeo-Johnson" if skew_val > 1.0 
            else "StandardScaler / None" if abs(skew_val) <= 0.5 
            else "Power Transform / Range Check"
        )
    }

def plot_univariate_distributions(df, columns, output_path=None):
    fig, axes = plt.subplots(len(columns), 2, figsize=(14, 3.8 * len(columns)), gridspec_kw={'width_ratios': [0.68, 0.32]})
    colors = ['#2b5c8f', '#e67e22', '#27ae60', '#8e44ad']
    
    for i, col in enumerate(columns):
        s = df[col].dropna()
        c = colors[i % len(colors)]
        st = compute_univariate_stats(s, col)
        
        ax_hist = axes[i, 0]
        discrete = True if col in ['Rating', 'Recommended IND'] else False
        sns.histplot(s, kde=not discrete, discrete=discrete, stat='density', color=c, ax=ax_hist, edgecolor='white', alpha=0.65)
        ax_hist.axvline(st['Mean'], color='#d9534f', linestyle='--', linewidth=2, label=f"Mean: {st['Mean']:.2f}")
        ax_hist.axvline(st['Median'], color='#2ecc71', linestyle='-', linewidth=2, label=f"Median: {st['Median']:.2f}")
        ax_hist.set_title(f"{col} - Distribution & Density (Skewness: {st['Skewness']:+.2f})", fontsize=11, weight='bold')
        ax_hist.set_ylabel("Density")
        ax_hist.grid(axis='both', linestyle=':', alpha=0.6)
        ax_hist.legend(loc='upper right', frameon=True)
        
        ax_box = axes[i, 1]
        sns.boxplot(y=s, color=c, ax=ax_box, fliersize=4, flierprops={'markerfacecolor': '#d9534f', 'alpha': 0.5})
        ax_box.set_title(f"{col} - Outlier Boxplot", fontsize=11, weight='bold')
        ax_box.set_ylabel(col)
        ax_box.grid(axis='y', linestyle=':', alpha=0.6)
        
        stat_box = f"Mean: {st['Mean']:.2f}\nMedian: {st['Median']:.2f}\nIQR: {st['IQR']:.2f}\nKurtosis: {st['Kurtosis']:+.2f}"
        ax_box.text(0.95, 0.95, stat_box, transform=ax_box.transAxes, fontsize=8.5, ha='right', va='top',
                    bbox=dict(boxstyle='round,pad=0.4', facecolor='white', edgecolor='#cccccc', alpha=0.9))
        
    plt.suptitle("Univariate Numerical Analysis Dashboard (Histograms, KDEs, Boxplots)", fontsize=13, weight='bold', y=1.008)
    plt.tight_layout()
    if output_path:
        plt.savefig(output_path, dpi=300, bbox_inches='tight')
    plt.show()

df = pd.read_csv("womens_clothing.csv")
plot_univariate_distributions(df, ['Age', 'Positive Feedback Count', 'Rating', 'Recommended IND'])
```

```python
def compute_category_entropy(series):
    clean_s = series.dropna().astype(str)
    probs = clean_s.value_counts(normalize=True).values
    k = len(probs)
    if k <= 1:
        return {"k_unique": k, "shannon_entropy": 0.0, "normalized_entropy": 0.0}
    shannon_ent = -float(np.sum(probs * np.log2(probs + 1e-12)))
    norm_ent = float(shannon_ent / np.log2(k))
    return {"k_unique": k, "shannon_entropy": round(shannon_ent, 3), "normalized_entropy": round(norm_ent, 3)}

def compute_mutual_information(feature_s, target_s):
    contingency = pd.crosstab(feature_s, target_s)
    p_xy = contingency.values / contingency.values.sum()
    p_x = p_xy.sum(axis=1)
    p_y = p_xy.sum(axis=0)
    expected = np.outer(p_x, p_y)
    mask = (p_xy > 0) & (expected > 0)
    return max(0.0, float(np.sum(p_xy[mask] * np.log2(p_xy[mask] / expected[mask]))))

class DynamicCategoricalEncoder:
    """
    Dynamically bins rare categories below a minimum frequency threshold into an 'Other' bucket
    and generates empirical frequency / proportion encodings.
    """
    def __init__(self, min_freq_pct=0.08):
        self.min_freq_pct = min_freq_pct
        self.frequent_categories_ = {}
        self.frequency_maps_ = {}
        
    def fit(self, df, columns):
        for col in columns:
            counts = df[col].value_counts(normalize=True)
            self.frequency_maps_[col] = counts.to_dict()
            self.frequent_categories_[col] = set(counts[counts >= self.min_freq_pct].index)
        return self
        
    def transform(self, df):
        df_out = df.copy()
        for col, frequent_set in self.frequent_categories_.items():
            df_out[f"{col}_binned"] = df_out[col].apply(lambda x: x if x in frequent_set else 'Other')
            df_out[f"{col}_freq_encoded"] = df_out[col].map(self.frequency_maps_[col]).fillna(0.0)
        return df_out

def plot_categorical_dashboard(df, target_col='Recommended IND', output_path=None):
    fig, axes = plt.subplots(2, 2, figsize=(14, 10))
    
    div_counts = df['Division Name'].value_counts()
    axes[0, 0].pie(div_counts.values, labels=div_counts.index, autopct='%1.1f%%', startangle=140,
                   colors=['#3498db', '#e74c3c', '#2ecc71'], wedgeprops=dict(width=0.45, edgecolor='white', linewidth=2))
    axes[0, 0].set_title("1. Division Name Market Share (Donut Chart)", fontsize=11, weight='bold')
    
    dept_counts = df['Department Name'].value_counts(ascending=True)
    axes[0, 1].barh(dept_counts.index, dept_counts.values, color='#2b5c8f', edgecolor='white', height=0.6)
    axes[0, 1].set_title("2. Department Review Volume Frequency", fontsize=11, weight='bold')
    axes[0, 1].set_xlabel("Review Count")
    axes[0, 1].grid(axis='x', linestyle='--', alpha=0.5)
    
    cross_tab = pd.crosstab(df['Department Name'], df[target_col], normalize='index') * 100
    cross_tab.plot(kind='bar', stacked=True, ax=axes[1, 0], color=['#e74c3c', '#2ecc71'], edgecolor='white', width=0.6)
    axes[1, 0].set_title("3. Recommendation Rate by Department (Stacked Bar)", fontsize=11, weight='bold')
    axes[1, 0].set_ylabel("Proportion (%)")
    axes[1, 0].set_xticklabels(cross_tab.index, rotation=30, ha='right')
    axes[1, 0].legend(["Not Recommended (0)", "Recommended (1)"], loc='lower right')
    axes[1, 0].grid(axis='y', linestyle='--', alpha=0.5)
    
    summary_data = [compute_category_entropy(df[c]) for c in ['Division Name', 'Department Name', 'Class Name', 'Title']]
    features = ['Division Name', 'Department Name', 'Class Name', 'Title']
    norm_ent = [s['normalized_entropy'] for s in summary_data]
    axes[1, 1].barh(features, norm_ent, color='#34495e', edgecolor='white', height=0.55)
    axes[1, 1].set_title("4. Normalized Shannon Entropy (Category Diversity)", fontsize=11, weight='bold')
    axes[1, 1].set_xlabel("Normalized Entropy (0 = Monopolized, 1 = Balanced)")
    axes[1, 1].set_xlim(0, 1.15)
    axes[1, 1].grid(axis='x', linestyle='--', alpha=0.5)
    
    plt.suptitle("Categorical Univariate Analysis Dashboard", fontsize=13, weight='bold', y=1.01)
    plt.tight_layout()
    if output_path:
        plt.savefig(output_path, dpi=300, bbox_inches='tight')
    plt.show()

df = pd.read_csv("womens_clothing.csv")
plot_categorical_dashboard(df, target_col='Recommended IND')
```
